# CIS 4400 Group Project: NYC 311 & Housing Code Violations Data Download

Python code to download NYC 311 Data using the Socrata API via the Sodapy library.

```python
# Downloading NYC OpenData 311 Data using Google Colab
# Likely need to install the sodapy module
%pip install sodapy
```

```python
import pandas as pd
from sodapy import Socrata
from google.colab import userdata

# The Host Name for the API endpoint. The https:// will be added by Sodapy
data_url='data.cityofnewyork.us'
# The data set at the API endpoint (311 data in this case)
data_set='erm2-nwe9'
# The app token created at data.cityofnewyork.us/profile/edit/developer_settings
# If you have the app token saved as a secret, access it here
app_token = userdata.get('nyc_open_data_token').strip()

# Create the client to point to the API endpoint
client = Socrata(data_url, app_token)
# Set the timeout to 600 seconds. If we get timeout errors, increase this value
client.timeout = 600
```

```python
# HEAT/HOT WATER - JANUARY 2026

start = 0
chunk_size = 2000
results_heat = []

where_clause = """
complaint_type = 'HEAT/HOT WATER'
AND created_date >= '2026-01-01T00:00:00'
AND created_date < '2026-02-01T00:00:00'
"""

# Count records
record_count_dict = client.get(
    data_set,
    where=where_clause,
    select="COUNT(*)"
)

record_count = int(record_count_dict[0]["COUNT"])

print(f"Total HEAT/HOT WATER records: {record_count}")

# Download records
while start < record_count:

    chunk = client.get(
        data_set,
        where=where_clause,
        offset=start,
        limit=chunk_size
    )

    results_heat.extend(chunk)

    start += chunk_size

    print(f"Downloaded {len(results_heat)} of {record_count}")
```

```python
# PLUMBING - JANUARY 2026

start = 0
chunk_size = 2000
results_plumbing = []

where_clause = """
complaint_type = 'PLUMBING'
AND created_date >= '2026-01-01T00:00:00'
AND created_date < '2026-02-01T00:00:00'
"""

# Count records
record_count_dict = client.get(
    data_set,
    where=where_clause,
    select="COUNT(*)"
)

record_count = int(record_count_dict[0]["COUNT"])

print(f"Total PLUMBING records: {record_count}")

# Download records
while start < record_count:

    chunk = client.get(
        data_set,
        where=where_clause,
        offset=start,
        limit=chunk_size
    )

    results_plumbing.extend(chunk)

    start += chunk_size

    print(f"Downloaded {len(results_plumbing)} of {record_count}")
```

```python
# UNSANITARY CONDITION - JANUARY 2026

start = 0
chunk_size = 2000
results_unsanitary = []

where_clause = """
complaint_type = 'UNSANITARY CONDITION'
AND created_date >= '2026-01-01T00:00:00'
AND created_date < '2026-02-01T00:00:00'
"""

# Count records
record_count_dict = client.get(
    data_set,
    where=where_clause,
    select="COUNT(*)"
)

record_count = int(record_count_dict[0]["COUNT"])

print(f"Total UNSANITARY CONDITION records: {record_count}")

# Download records
while start < record_count:

    chunk = client.get(
        data_set,
        where=where_clause,
        offset=start,
        limit=chunk_size
    )

    results_unsanitary.extend(chunk)

    start += chunk_size

    print(f"Downloaded {len(results_unsanitary)} of {record_count}")
```

```python
# Convert each API result into a DataFrame

df_heat = pd.DataFrame.from_records(results_heat)

df_plumbing = pd.DataFrame.from_records(results_plumbing)

df_unsanitary = pd.DataFrame.from_records(results_unsanitary)


# Combine all three complaint types into one 311 DataFrame

df_311 = pd.concat(
    [df_heat, df_plumbing, df_unsanitary],
    ignore_index=True
)
```

```python
# Save the combined DataFrame as a CSV
df_311.to_csv(
    "nyc_311_january_2026.csv",
    index=False
)
```

```python
# The Host Name for the API endpoint. The https:// will be added by Sodapy
housing_data_url = 'data.cityofnewyork.us'

# Housing Maintenance Code Violations dataset
housing_data_set = 'wvxf-dwi5'

# Access app token
housing_app_token = userdata.get('nyc_open_data_token').strip()

# Create a separate client for the Housing dataset
housing_client = Socrata(housing_data_url, housing_app_token)

# Set timeout
housing_client.timeout = 600
```

```python
# Start at record 0 of the results
housing_start = 0

# Fetch 2000 rows at a time
housing_chunk_size = 2000

# Empty out our result list
housing_results = []

# Filter for violations with an inspection date in January 2026
housing_where_clause = """
inspectiondate >= '2026-01-01T00:00:00'
AND inspectiondate < '2026-02-01T00:00:00'
"""

# See how many violation records there are
housing_record_count_dict = housing_client.get(
    housing_data_set,
    where=housing_where_clause,
    select="COUNT(*)"
)

# Save as an integer
housing_record_count = int(
    housing_record_count_dict[0]['COUNT']
)

print(f"Total January 2026 HPD violation records: {housing_record_count}")

# Loop until we have fetched all of the records
while housing_start < housing_record_count:

    # Fetch the set of records starting at housing_start
    housing_results.extend(
        housing_client.get(
            housing_data_set,
            where=housing_where_clause,
            offset=housing_start,
            limit=housing_chunk_size
        )
    )

    # Move up the starting record for the next chunk
    housing_start = housing_start + housing_chunk_size

    # Give some feedback
    print(
        f"Downloaded {len(housing_results)} "
        f"of {housing_record_count}"
    )
```
